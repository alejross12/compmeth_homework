```python
Rsun = 696340000    # meters
sigT = 6.652e-25
def ne(r):
    # r in meters!
    return (2.5 * 10**(26)) * np.exp(-r / (0.096 * Rsun))

r = np.linspace(0, 0.9*Rsun, 1000)
y = ne(r)
p = ne(r) * sigT * 50
l = 1 / (ne(r) * sigT)
stepsize = (10*l / 100)     # 10 mean-free-paths, cm -> m

# plt.plot(r, y)
# plt.plot(r, (10*l / 100))
plt.plot(r, y)
# plt.xlim(0, 1.8e8)
```
